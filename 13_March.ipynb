```python
#Q1 Write a program to find sum of digits
def SumOfDigits(n,res):
    if(n == 0):
        return res
    rem = n%10
    return SumOfDigits(n//10,res+rem)
n = 12345
print(SumOfDigits(n,0))
```

```python
#Q2 factorial 
def SumOfDigits(n):
    if(n == 0):
        return 1
    return n*SumOfDigits(n-1)
n = 5
print(SumOfDigits(5))
```

```python
#Q3 Find Fabonacci
def Fabonacci(n):
    if(n == 1 or n == 0):
        return 1
    return Fabonacci(n-1) + Fabonacci(n-2)

n = 10
print(Fabonacci(n))
```

```python
#Q4 Reverse a string using recursion
def Reverse(str):
    if(str == ""):
        return str
    return str[-1] + Reverse(str[:-1])

str = "asdf"
print(Reverse(str))
```

```python
#Q5 Check if the function is Palindrome
def checkPalindrome(sentence,left,right):
    if(left >= right):
        return True
    if(sentence[left] != sentence[right]):
        return False
    return checkPalindrome(sentence,left+1,right-1)

x = "abcba"
print(checkPalindrome(x,0,len(x)-1))
```
